# Map Function  in Python 
Map function is basically to apply function to all the item in the iterable one by one very much usefull in the list comprehensive.

Map function takes two input  
```map(function,iterable)```  

It's basicaly used in various places that can be very effective for iteration.can be used for list comprehension,lambda and many more 

````python
def number_square(x):
    return x**2
number=[1,2,3,4,5,6,8]
#map function
square=map(number_square,number)
print(square)
# The data is assingned in somewhere in the memory location.
````

````python
# If you want to get the output after performing the expression you can do this.
print(list(map(number_square,number)))
````

Now we use the lambda function in the map function.

````python
# lambda function 
number2=[1,2,3,4,5,56,7,7]
print(list(map(lambda x:x**2,number2)))
````

Here we can also perform iteration and converation of one data type to another data type.

````python
# Conversion from the string to int
number2=['1','2','3','4']
number3=list(map(int,number2))
print(number3)
````

Converting the lower case into the upper case.

````python
# Conversion form lower case to upper case
list2=['apple','banana','cake']
list3=list(map(str.upper,list2))

print(list3)
````

We can also use this in the dictinaries.

````python
def peoplename(people):
    return people['name']
company=[{'name':'himanshu','class':11},
        {'name':'yuvraj','class':12},
        {'name':'manish','class':12}]
list4=list(map(peoplename,company))
print(list4)
````
