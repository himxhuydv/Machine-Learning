# Filter Function in Python 


Filter function perform a iteration through iterable on each element based on the applied condition and filter out the value which are true based on the applied condition.it's filter out the value on the basis of the applied condition.

```python
# With function
def even(x):
    x%2==0
    return True # This literally 
list2=[12,3,32,2,42,4422,24,21]
filter_value=list(filter(even,list2))
print(filter_value)
```

```python
# Lambda function
filter_new_value=list(filter(lambda x:x%2==0,list2))
print(filter_new_value)
```

```python
# Lambda function with multiple condition
list2=[12,3,32,2,42,4422,24,21]
filter_new_valus=list(filter(lambda x:x%2==0 and x>50,list2))
print(filter_new_valus)
```

```python
# Filter function with dictionaries 
def name_dict(people):
    return people['class']>16
data1={'name':'himanshu','class':12},{'name1':'utsav','class':12},{'name2':'himanshs','class':23}
value_list=list(filter(name_dict,data1))
print(value_list)
```
