Temperature Converter 

In the temperature converter we basically convert the celcius to farhrenhiet and farhrenhiet to celicus these two are the only way.

My way of solving

```python
def celcius_to_fahrenhiet(Celcius):
    """This function convert the celcius to fahrenhiet"""
    fahrenhiet=Celcius*9/5+32
    return fahrenhiet
def fahrenhiet_to_celcius(farh):
    """This function convert fshrenhiet to celcius"""
    celcius=farh-32*5/9
    return celcius


    
```

```python
# Calling of function of celcius to fahrenhiet 
celcius_to_fahrenhiet(23)
```

```python
# Calling of function of fahrenhiet to celcius
fahrenhiet_to_celcius(23)
```

# smart way of solving

```python
def temp_convert(temp,unit):
    """This function convert the fahrenhiet to celicus and celcius to fahrenhiet in smart way by taking unit in string and checking that F or C."""
    if unit=='C':
        print(temp*9/5+32)
    elif unit=='F':
        print(temp-32*5/9)
    else:
        print("Error kindly input In Fahrenhiet and Celcius")
```

```python
# Calling of the temp_convert function():
temp_convert(23,"C")
```

Check if a string is a palindrome.

```python
def check_if_palindrom(word):
    stage=''
    words+=word.lower().replace(" ","")
    for i in words[::-1]:
        stage+=i
    if stage==words:
        print('Palindrome')
    else:
        print('Not Palindrome')
Enter=['madam is a sus kind of wow']
check_if_palindrom(Enter)
```
