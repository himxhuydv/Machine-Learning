# Positional Arguments 

```python
def variable_length(*argu):
    for value in argu:
        print(value)
# Calling of the function.....
print(variable_length(1,2,3,4,4,5,6,7,8,"himxhuydv"))
```

# Keyword Arguments
used where the element in the form of key value pair.... that's where we write it.

```python
def keyword_arg(**kwargs):
    for key,value in kwargs.items(): # in the dict we use the item for the acessing 
        print(f"{key}:{value}")
keyword_arg(Name="himanshu",age=23,country=23)
```

# Impliment them togther positional arguments and positional arguments

```python
def post_key(*args,**kwargs):
    for val in args:
        print(val)
    for key,value in kwargs.items():
        print(key,value)
post_key(1,2,3,4,5,5,6,"himxhuydv",name="himanshu",surname="yadav",)
```
