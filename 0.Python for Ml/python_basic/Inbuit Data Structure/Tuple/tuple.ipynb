# Slicing in the nested tuple

```python
#nested tuple 
tu=((1,2,3,42,1),(221,3213,42,1,"hello"),("hello2","you","im"))
print(tu[0]) # first of the tuple will be printed 
print(tu[0][1]) # you can easily print out the desired element from the tuple
print(tu[2][1])
```
