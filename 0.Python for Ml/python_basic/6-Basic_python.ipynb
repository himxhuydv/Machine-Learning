# List Comprehension:
generally used for short and powerfull way of the list creation and more importantly in single line.


```python
l_com=[x**2 for x in range(0,20)]
print(l_com)
```

 
**List Comprehension with basic syntax**  
List [expression and for item in iterable]  
**List Comprehension with conditional statement**  
List [expression and for item in iterable condition]  

```python
# List comprehension with the basic syntax
List_co=[num**2 for num in range(0,10)]
print(List_co)
# List Comprehension with the basic syntax and the conditional statement
l1=[num for num in range(0,20)if num%2==0]
print(l1)
```
