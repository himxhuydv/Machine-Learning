# Creation of Array 


basically array is used in the Numpy.


1 D Array: basically mean the array is single dimension.

```python
import numpy as np
arr_1d=np.array([1,2,3,4,5,6])
print("1_D array",arr_1d)
```

2 D Array: basically mean the array is having two dimension -rows and coloumn.

```python
import numpy as np 
arr_2d=np.array([[1,2,3,4,5],[1,3,4,5]])
print("2 D array",arr_2d)
```

Special array:this basically mean the array is following any specific patterns.

Zero Array

```python
import numpy as np
arr_3d=np.zeros((2,3),int)
print("Zeros array",arr_3d)
```

Once Array

```python
import numpy as np
arr_4d=np.ones((2,4),int)
print(arr_4d)
```

LineSpace Array:basically mean the there is (start: end : step)this kind of way to print.

```python
import numpy as np 
arr_5d=np.linspace(0,10,2)
print(arr_5d)
```
